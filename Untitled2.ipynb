```python
import numpy as np
import pandas as pd
import time
from scipy import stats

# =====================================================================
# 1. PARAMÈTRES DE LA SIMULATION
# =====================================================================
NUM_MODULES = 12
BUFFER_CAPACITY = 10
TOTAL_TIME_STEPS = 500

DELTA_T_SEC = 1.0  # 1 pas de temps = 1 seconde
DELTA_T_HOURS = DELTA_T_SEC / 3600.0  # Conversion en heures

P_ACTIVE_MIN, P_ACTIVE_MAX = 3.5, 6.0
P_IDLE_MIN, P_IDLE_MAX = 0.4, 0.9

# Poids statiques par défaut pour les baselines
ALPHA_DEFAULT = 0.40
BETA_DEFAULT = 1.20
OMEGA_PENALTY = 1e6

FAILURE_TIMINGS = [100, 250, 380]
FAILURE_DURATION = 30

# =====================================================================
# 2. ENVIRONNEMENT RMS
# =====================================================================
class RMSEnvironment:
    def __init__(self, seed):
        self.rng = np.random.RandomState(seed)
        self.num_modules = NUM_MODULES
        self.buffer_capacity = BUFFER_CAPACITY

        self.p_active = self.rng.uniform(P_ACTIVE_MIN, P_ACTIVE_MAX, size=self.num_modules)
        self.p_idle = self.rng.uniform(P_IDLE_MIN, P_IDLE_MAX, size=self.num_modules)
        self.processing_times = self.rng.randint(4, 10, size=self.num_modules)

        self.reset()

    def reset(self):
        self.time_step = 0
        self.buffers = np.zeros(self.num_modules, dtype=int)
        self.busy_until = np.zeros(self.num_modules, dtype=int)
        self.failures = np.ones(self.num_modules, dtype=int)
        self.cumulative_energy_kwh = 0.0
        self.cumulative_overflow = 0.0
        return self.get_state()

    def update_failures(self):
        self.failures[:] = 1
        for f_time in FAILURE_TIMINGS:
            if f_time <= self.time_step < f_time + FAILURE_DURATION:
                self.failures[[2, 5, 9]] = 0

    def step(self, routing_action):
        self.update_failures()

        if routing_action is not None and 0 <= routing_action < self.num_modules:
            self.buffers[routing_action] += 1

        for m in range(self.num_modules):
            if self.failures[m] == 1 and self.time_step >= self.busy_until[m]:
                if self.buffers[m] > 0:
                    self.buffers[m] -= 1
                    self.busy_until[m] = self.time_step + self.processing_times[m]

        power_t = np.where(
            (self.time_step < self.busy_until) & (self.failures == 1),
            self.p_active,
            self.p_idle
        )

        step_energy_kwh = np.sum(power_t) * DELTA_T_HOURS
        self.cumulative_energy_kwh += step_energy_kwh

        step_overflow = np.sum(np.maximum(0, self.buffers - self.buffer_capacity))
        if routing_action is not None and self.failures[routing_action] == 0:
            step_overflow += 5.0

        self.cumulative_overflow += step_overflow
        self.time_step += 1

        return self.get_state()

    def get_state(self):
        return {
            'time_step': self.time_step,
            'buffers': self.buffers.copy(),
            'failures': self.failures.copy(),
            'p_active': self.p_active.copy(),
            'p_idle': self.p_idle.copy()
        }

# =====================================================================
# 3. STRATÉGIES DE ROUTAGE
# =====================================================================
def route_static_no_redirect(state):
    return 2

def route_static_with_redirect(state):
    candidates = [2, 3, 4]
    for c in candidates:
        if state['failures'][c] == 1:
            return c
    return candidates[0]

def route_shortest_queue(state):
    active_modules = np.where(state['failures'] == 1)[0]
    if len(active_modules) == 0:
        return 0
    return active_modules[np.argmin(state['buffers'][active_modules])]

def route_earliest_completion_time(state):
    active_modules = np.where(state['failures'] == 1)[0]
    if len(active_modules) == 0:
        return 0
    estimated_times = state['buffers'][active_modules] * 6.0
    return active_modules[np.argmin(estimated_times)]

def route_round_robin(state, step_counter):
    active_modules = np.where(state['failures'] == 1)[0]
    if len(active_modules) == 0:
        return 0
    return active_modules[step_counter % len(active_modules)]

def route_flat_cost(state, alpha=ALPHA_DEFAULT, beta=BETA_DEFAULT):
    """ Baseline Flat : Poids fixes statiques quel que soit l'état du système """
    costs = []
    p_max = P_ACTIVE_MAX
    b_max = BUFFER_CAPACITY

    for m in range(NUM_MODULES):
        if state['failures'][m] == 0:
            costs.append(OMEGA_PENALTY)
        else:
            e_norm = state['p_active'][m] / p_max
            q_norm = state['buffers'][m] / b_max
            cost = alpha * e_norm + beta * q_norm
            costs.append(cost)
    return int(np.argmin(costs))

def route_proposed_mod_sbsg(state, alpha_base=ALPHA_DEFAULT, beta_base=BETA_DEFAULT):
    """
    Méthode Proposée Mod-SbSG :
    Le Leader observe l'état global (pannes, congestion) et ajuste dynamiquement
    les poids alpha_t et beta_t imposés aux Followers.
    """
    costs = []
    p_max = P_ACTIVE_MAX
    b_max = BUFFER_CAPACITY

    global_congestion = np.mean(state['buffers']) / b_max
    has_active_failure = np.any(state['failures'] == 0)

    # STRATÉGIE DU LEADER BASÉE SUR L'ÉTAT :
    if has_active_failure or global_congestion > 0.35:
        # En cas de crise (panne ou congestion) : priorité au désengorgement
        alpha_t = alpha_base * 0.25  # Poids énergie réduit
        beta_t = beta_base * 2.50   # Poids file d'attente amplifié
    else:
        # Régime nominal : priorité à l'économie d'énergie
        alpha_t = alpha_base * 1.50  # Poids énergie augmenté
        beta_t = beta_base * 0.70   # Poids file toléré

    for m in range(NUM_MODULES):
        if state['failures'][m] == 0:
            costs.append(OMEGA_PENALTY)
        else:
            e_norm = state['p_active'][m] / p_max
            q_norm = state['buffers'][m] / b_max
            cost = alpha_t * e_norm + beta_t * q_norm
            costs.append(cost)

    return int(np.argmin(costs))

# =====================================================================
# 4. SIMULATION ET EVALUATION STATISTIQUE MULTI-GRAINES (N=30)
# =====================================================================
def run_experiments(num_seeds=30):
    seeds = range(42, 42 + num_seeds)
    algorithms = {
        'Static (No-Redirect)': lambda st, step: route_static_no_redirect(st),
        'Static (Redirect)': lambda st, step: route_static_with_redirect(st),
        'Shortest Queue (SQ)': lambda st, step: route_shortest_queue(st),
        'Earliest Completion (ECT)': lambda st, step: route_earliest_completion_time(st),
        'Round-Robin (RR)': lambda st, step: route_round_robin(st, step),
        'Flat Non-Hierarchical': lambda st, step: route_flat_cost(st),
        'Proposed Mod-SbSG': lambda st, step: route_proposed_mod_sbsg(st),
        'Ablation (No Energy, alpha=0)': lambda st, step: route_proposed_mod_sbsg(st, alpha_base=0.0),
        'Ablation (No Congestion, beta=0)': lambda st, step: route_proposed_mod_sbsg(st, beta_base=0.0),
    }

    results = {alg: {'energy': [], 'overflow': [], 'latency_ms': []} for alg in algorithms}

    for seed in seeds:
        for alg_name, alg_fn in algorithms.items():
            env = RMSEnvironment(seed=seed)
            state = env.get_state()
            latencies = []

            for t in range(TOTAL_TIME_STEPS):
                t_start = time.perf_counter()
                action = alg_fn(state, t)
                t_end = time.perf_counter()

                latencies.append((t_end - t_start) * 1000.0)
                state = env.step(action)

            results[alg_name]['energy'].append(env.cumulative_energy_kwh)
            results[alg_name]['overflow'].append(env.cumulative_overflow)
            results[alg_name]['latency_ms'].append(np.mean(latencies))

    return results

def summarize_and_test(results):
    summary = []
    proposed_energy = results['Proposed Mod-SbSG']['energy']
    proposed_overflow = results['Proposed Mod-SbSG']['overflow']

    for alg_name, metrics in results.items():
        e_mean, e_std = np.mean(metrics['energy']), np.std(metrics['energy'])
        o_mean, o_std = np.mean(metrics['overflow']), np.std(metrics['overflow'])
        l_mean, l_std = np.mean(metrics['latency_ms']), np.std(metrics['latency_ms'])

        if alg_name != 'Proposed Mod-SbSG':
            diff_e = np.array(metrics['energy']) - np.array(proposed_energy)
            if np.all(diff_e == 0):
                p_val_e = 1.0
            else:
                _, p_val_e = stats.wilcoxon(metrics['energy'], proposed_energy)

            diff_o = np.array(metrics['overflow']) - np.array(proposed_overflow)
            if np.all(diff_o == 0):
                p_val_o = 1.0
            else:
                _, p_val_o = stats.wilcoxon(metrics['overflow'], proposed_overflow)
        else:
            p_val_e, p_val_o = 1.0, 1.0

        summary.append({
            'Algorithm': alg_name,
            'Energy (kWh)': f"{e_mean:.2f} ± {e_std:.2f}",
            'Overflow': f"{o_mean:.2f} ± {o_std:.2f}",
            'Latency (ms)': f"{l_mean:.3f} ± {l_std:.3f}",
            'p-val Energy': f"{p_val_e:.4f}",
            'p-val Overflow': f"{p_val_o:.4f}"
        })

    return pd.DataFrame(summary)

if __name__ == "__main__":
    print("=== EXÉCUTION DE LA SIMULATION CORRIGÉE MOD-SBSG (N=30) ===")
    results_data = run_experiments(num_seeds=30)
    df_res = summarize_and_test(results_data)
    print(df_res.to_string(index=False))
```
